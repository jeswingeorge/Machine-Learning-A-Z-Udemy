## What decision is this model supporting?

Eg: Are you ranking cases where your team can review the highest risk for you?  Or, are you making a binary decision at a fixed threshold? Are false positives relatively harmless or do they create real operational cost? Are positive classes rare?

These questions help to decide which metric to use in your evaluation?

ROC-AUC and PR-AUC tell different things about the predictions made by the model, and they are not interchangeable.

> Classification models usually assign a score (confidence/probability) to each example. To make a prediction, you choose a threshold.

### Score, Threshold and Probability

**Score**  reflects how likely it is to belong to +ve class.  To turn a score into actual prediction we choose a **threshold**. Above that threshold you predict positive below it you predict negative.  

![](images/Y2.png)

And that Threshold matters, if you set it high the model becomes too conservative about calling something **positive**.   
![](images/Y3.png)

If we lower the threshold a lot, it becomes more permissive.  
![](images/Y4.png)

So when we evaluate a classifier one important question is not just how it performs at one threshold but how its behaviour changes as we move the threshold up and down. That's the context for **ROC-AUC**.






#### What ROC-AUC tell us?

ROC AUC measures how well the model separates positives from negatives across all thresholds. It is threshold-independent and useful for general ranking quality.

ROC plots:  An ROC (Receiver Operating Characteristic) curve is a graph that shows the performance of a binary classification model across all possible classification thresholds. 

* **Y-Axis (True Positive Rate / Sensitivity / Recall)**: Measures the proportion of actual positive cases correctly identified by the model $$\left(\frac{\text{True Positives}}{\text{True Positives} + \text{False Negatives}}\right)$$
* **X-Axis (False Positive Rate / 1 - Specificity)**: Measures the proportion of actual negative cases incorrectly flagged as positive $$\left(\frac{\text{False Positives}}{\text{False Positives} + \text{True Negatives}}\right)$$  
* Thresholds: Each point on the curve represents a different decision threshold (the probability cut-off used to turn a predicted score into a class, usually default 0.5)

ROC-AUC is good for early stage model development and model comparison.

The catch: ROC AUC can be misleading in imbalanced settings because it uses the false positive rate, which can look deceptively small even when there are many raw false positives.

# Precision-Recall AUC (PR-AUC) Curve 

It focuses on the quality of positive predictions. It is more sensitive to operational reality when the positive class is rare.

A precision-recall curve is a graph that shows the trade-off between precision (correct positive predictions) and recall (actual positive cases found) for a binary classification model across different decision thresholds.  
## Key Definitions

* Precision: The ratio of true positives to all predicted positives ($TP / [TP + FP]$). It measures how reliable a positive prediction is.
* Recall (Sensitivity): The ratio of true positives to all actual positives ($TP / [TP + FN]$). It measures how many of the actual positive items the model found.
* Axes: Recall is plotted on the x-axis, and Precision is plotted on the y-axis. 

## Why Use a Precision-Recall Curve?

* Imbalanced Data: It works best when one class is rare or dominant (like fraud detection or disease screening). 
* Focus on the Positive Class: Unlike ROC curves, it ignores true negatives, making it much more sensitive to false positives when the negative class is huge. 
* Threshold Tuning: Every point on the curve represents a different probability cut-off, helping you pick the best balance for your specific application.  
* Evaluation (AUC-PR): The area under the precision-recall curve summarizes overall performance; a score closer to 1.0 means a better model. You can compute this using tools like the [Scikit-Learn Precision Recall Curve Module](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_recall_curve.html).

Keep in mind ROC-AUC is not wrong but its answering a different question. 



A subtle but important mistake is stopping at AUC entirely.

In prodution, nobody deploys all thresholds, you deploy one threshold and may be a few operating points.

![](images/Y6.png)

So even if ROC-AUC or PR-AUC is your summary metric, you still need threshold level evaluation.  

Questions from decision evaluation which corresponds back to model evaluation:
- What is Precision at the top 1%?  
- What Recall you get at the review capacity your ops team can handle?
- How many False positives per day?
- What is Downstream cost?

So at this point you stop asking if the model is statistically better and start asking if the operating point is systemically viable.
